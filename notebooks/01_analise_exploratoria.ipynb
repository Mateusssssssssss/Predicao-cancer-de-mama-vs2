# 1. Análise exploratória
Carregamento e validação do dataset, distribuição das classes, correlações e gráficos. Execute a partir da raiz do repositório.

```python
from breast_cancer.config import DATA_PATH, FIGURE_DIR
from breast_cancer.data import load_data, clean_data
from breast_cancer.visualization import save_eda_figures

raw = load_data(DATA_PATH)
data = clean_data(raw)
print(f'Dimensões: {data.shape}')
print(f'Duplicados: {data.duplicated().sum()}')
display(data.isna().sum().sort_values(ascending=False).head(10))
display(data['diagnosis'].value_counts().rename(index={'B': 'Benigno', 'M': 'Maligno'}))
```

```python
save_eda_figures(raw, FIGURE_DIR)
print(f'Figuras salvas em {FIGURE_DIR}')
```
